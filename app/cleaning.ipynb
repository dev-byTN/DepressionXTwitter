<h3>Import useful librairies and load csv file created previously</h3>

```python
import pandas as pd
import numpy as np
```

```python
dt = pd.read_json("ressource/fetchedTweets.json")
dt
```

<h3>Let's remove the unnecessary text format</h3>

```python
import emoji
```

```python
dt["tweet"] = dt['tweet'].str.replace("\n", "")
```

```python
dt["tweet"] = dt["tweet"].apply(lambda x : emoji.replace_emoji(x, ""))
```

```python
dt["tweet"] = dt["tweet"].str.replace(r"http\S+|@\S+|#\S+", "", regex=True)
dt
```

<h3>Now le's work on the date and time column</h3>

```python
dt.insert(3,"day", 0)
dt.insert(4, "month", 0)
dt.insert(5,"year", 0)
dt.insert(6,"hour",0)
dt.insert(9, "creation",0)
```

```python
# All the dates are written the same way : 'Day Month Day(digit)' & at the end the year
# So the format is the same : 3 letters for the day + " " + 3 letters for the month + " " + 2 digits for the day
# We want to export the year for the date to the year column created previously
```

```python
dt['date'] = dt['date'].dt.strftime('%Y-%m-%d %H:%M:%S %Z')  #convert a datetime into a str
```

```python
dt.date = dt.date.str.replace(" UTC", "")
```

```python
dt["hour"] = dt['date'].str[11:16]
dt["year"] = dt["date"].str[0:4]
dt['day'] = dt['date'].str[8:10]
dt.month = dt.date.str[5:7]

dt
```

```python
dt.drop("date", axis="columns", inplace=True)
```

```python
dt.dtypes
```

```python
dt.creation = dt.createdAt.str.slice(start=4, stop=10) + " " + dt.createdAt.str.slice(-4)
```

```python
dt.drop("createdAt", axis="columns", inplace=True)
dt
```

```python
dt = dt.astype ({
        "day":int,
        "year":int,
        "month":int
})
```

```python
dt.to_json("ressource/cleanedTweets.json", orient="records")
```
